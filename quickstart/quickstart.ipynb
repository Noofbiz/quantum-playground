# Quickstart
This file follows the quickstart for IBM's Qiskit that can be found [here](https://quantum.cloud.ibm.com/docs/en/guides/quick-start)

First we start by building a bell state of two entagled qbits:

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()

sampler = StatevectorSampler()
result = sampler.run([qc], shots=1024).result()
print(result[0].data.meas.get_counts())
```
